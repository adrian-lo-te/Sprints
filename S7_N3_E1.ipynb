```python
import string
import numpy as np
```

La primera funció és tipus_caracter(). Aquesta funció rep un generador aleatori i diverses opcions booleanes per incloure majúscules, minúscules, números i signes. Primer he creat una llista buida i, mitjançant estructures condicionals if, comprovo quins tipus de caràcters s'han de seleccionar; per a cadascun, utilitzo .choice() de rng combinat amb els mòduls de string per afegir un caràcter aleatori de cada categoria a la llista. Finalment, es retorna aquesta llista.

```python
def tipus_caracter(rng, majuscules=True, minuscules=True, numeros=True, signes=True):

    caracters = []

    if majuscules:
        caracters.append(rng.choice(list(string.ascii_uppercase)))
    if minuscules:
        caracters.append(rng.choice(list(string.ascii_lowercase)))
    if numeros:
        caracters.append(rng.choice(list(string.digits)))
    if signes:
        caracters.append(rng.choice(list(string.punctuation)))

    return caracters
```

La següent funció és longitud_contrasena(), que s'encarrega d'assegurar la longitud total de la contrasenya. Primer comprova si la longitud total és menor que la llista actual de caràcters per ajustar-la si cal. Després, construeix una cadena de prova amb els conjunts de caràcters permesos segons els paràmetres actius, establint un valor per defecte si no hi ha cap opció seleccionada. A continuació, calcula els caràcters que falten i, si n'hi ha de pendents, utilitza rng.choice() per afegir els caràcters extra necessaris a la llista abans de retornar-la

```python
def longitud_contrasena(caracters_actuals, longitud_total, rng, majuscules=True, minuscules=True, numeros=True, signes=True):

    if longitud_total < len(caracters_actuals):
        longitud_total = len(caracters_actuals)

    prova = ""
    if majuscules:
        prova += string.ascii_uppercase
    if minuscules:
        prova += string.ascii_lowercase
    if numeros:
        prova += string.digits
    if signes:
        prova += string.punctuation


    if not prova:
        prova = string.ascii_letters + string.digits

    llista_prova = list(prova)
    caracters_faltans = longitud_total - len(caracters_actuals)

    if caracters_faltans > 0:
        caracters_extra = rng.choice(llista_prova, size=caracters_faltans).tolist()
        return caracters_actuals + caracters_extra

    return caracters_actuals
```

Aquesta funció anomenada permutar rep la llista de caràcters i un generador aleatori. Utilitzo .permutation() de rng per barrejar de manera aleatòria els elements de la llista i, finalment, uneixo tots els caràcters amb .join() per convertir-los en una sola cadena de text que es retorna

```python
def permutar(llista_caracters, rng):

    llista_permutada = rng.permutation(llista_caracters)
    return "".join(llista_permutada)
```

Per finalitzar, aquesta funció principal anomenada generar_contrasena() integra totes les anteriors. Inicialitza el generador aleatori amb np.random.default_rng() i crida successivament a les funcions tipus_caracter(), longitud_contrasena() i permutar() per generar, completar i barrejar finalment la contrasenya que es retorna

```python
def generar_contrasena(longitud=12, majuscules=True, minuscules=True, numeros=True, signes=True):

    rng = np.random.default_rng()

    caracters = tipus_caracter(rng, majuscules, minuscules, numeros, signes)
    llista_completa = longitud_contrasena(caracters, longitud, rng, majuscules, minuscules, numeros, signes)
    contrasena_final = permutar(llista_completa, rng)

    return contrasena_final
```

```python
pwd = generar_contrasena(longitud=12, majuscules=True, minuscules=True, numeros=True, signes=True)
print(f"Contraseña generada: {pwd}")
```
