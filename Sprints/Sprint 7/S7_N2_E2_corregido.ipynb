```python
import pandas as pd
```

La primera funció és seguro_df(), que rep un diccionari i el transforma en un data frame utilitzant .from_dict() amb orient="index". Després, utilitzo .reset_index() amb el paràmetre names="codi" per convertir l'índex del diccionari en una columna nova.

```python
def seguro_df(dicc):
  taula_seguro = pd.DataFrame.from_dict(dicc, orient="index")
  taula_seguro_nivells = taula_seguro.reset_index(names="codi")
  return taula_seguro_nivells
```

Aquesta funcio transforma la llista a data frame.

```python
def clients_df(llista):
  client_s = pd.DataFrame(llista)
  return client_s
```

Aquesta funció rep la taula de nivells i la taula de clients, i les junta utilitzant pd.merge() amb un tipus d'unió esquerra (how="left") a partir de la columna en comú "codi".

```python
def juntar_df(taula_nivells, taula_clients):
  nova_taula = pd.merge(taula_clients, taula_nivells, on = "codi", how = "left")
  return nova_taula
```

Aquesta funció crea una nova columna derivada al DataFrame anomenada "probabilitat_frau", multiplicant els valors de la columna "fraude" per 100 per expressar-ho en format percentual.

```python
def probabilitat_frau(taula):
  taula["probabilitat_frau"] = taula["fraude"] * 100
  return taula
```

Aquesta funció fa una operació similar a l'anterior per calcular el "preu_final", multiplicant el "preu_base" per un factor d'ajust que depèn de l'edat de cada client.

```python
def calcular_preu(taula):
  taula["preu_final"] = taula["preu_base"] * (1 + (taula["edat"]/100))
  return taula
```

Aquesta funció ordena primer per probabilitat de frau (descendent) i després per preu final (descendent)

```python
def ordenar_clients(taula):

    taula_ordenada = taula.sort_values(by=["probabilitat_frau", "preu_final"], ascending=[False, False])
    return taula_ordenada
```

Aquesta funció busca els registres amb valors màxims utilitzant .idxmax() tant per a la probabilitat de frau com per al preu final. Després, construeix i imprimeix per pantalla un apartat d'anàlisi de risc detallat amb missatges personalitzats

```python
def analisi(taula):

  client_max_frau = taula.loc[taula["probabilitat_frau"].idxmax()]
  client_max_preu = taula.loc[taula["preu_final"].idxmax()]

  print("=" * 40)
  print("       APARTAT D'ANÀLISI DE RISC")
  print("=" * 40)
  print(
      f" Client amb més probabilitat de frau: {client_max_frau['nom']}"
      f" ({client_max_frau['probabilitat_frau']}%)"
  )
  print(
      f" Client amb el preu final més alt: {client_max_preu['nom']}"
      f" ({client_max_preu['preu_final']} €)"
  )
  print(
      f" Prioritari de revisar manualment: {client_max_frau['nom']} (per"
      " presentar el nivell de risc crític i major probabilitat de frau)"
  )
  print("=" * 40)
```

processar() processa totes les funcions i retorna la taula i les respostes.

```python
def processar(dicc, llista):
  taula_nivells = seguro_df(dicc)
  taula_clients = clients_df(llista)
  taula_junta = juntar_df(taula_nivells, taula_clients)
  taula_prob = probabilitat_frau(taula_junta)
  taula_total = calcular_preu(taula_prob)
  taula_ordenada = ordenar_clients(taula_total)

  analisis = analisi(taula_ordenada)
  return display(taula_ordenada), analisis
```

```python
clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}

processar(nivells_seguro, clients)
```
