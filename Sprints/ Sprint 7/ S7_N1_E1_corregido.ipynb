```python
import re
```

La primera funció és validar_dades(), que comprova que els valors introduïts estiguin dins dels rangs permesos. Utilitzo estructures condicionals per verificar que l'experiència no sigui negativa i que els nivells de català, excel i formació es trobin estrictament entre 0 i 3.

```python
def validar_dades(exp, cat, exc, form):
  if exp < 0:
    return False
  if cat not in [0, 1, 2, 3]:
    return False
  if exc not in [0, 1, 2, 3]:
    return False
  if form not in [0, 1, 2, 3]:
    return False
  else:
    return True
```

La següent funció és obtenir_dades(), encarregada de gestionar l'entrada de dades de cada candidat. Primer demano el nom i utilitzo .strip() per netejar espais, validant amb un bucle que no estigui buit. Després, sol·licito el telèfon i comprovo amb una expressió regular (re.fullmatch) que tingui exactament 9 dígits. Finalment, mitjançant un bloc try-except, demano els anys d'experiència i els nivells, cridant a validar_dades() per assegurar-me que tot és correcte abans de retornar un diccionari amb les dades o None si hi ha errors.

```python
def obtenir_dades(candidat):
    print(f"\n--- Registre Candidat {candidat} ---")

    nom = input("Nom i Cognoms: ").strip()
    while not nom:
        print("El nom no pot estar buit.")
        nom = input("Nom i Cognoms: ").strip()

    telefon = input("Telèfon de contacte (9 dígits): ").strip()
    while not re.fullmatch(r'\d{9}', telefon):
        print("Error: El telèfon ha de tenir exactament 9 dígits numèrics.")
        telefon = input("Telèfon de contacte (9 dígits): ").strip()

    while True:
        try:
            experiencia = int(input("Anys d'experiència: "))
            catala = int(input("Nivell de català (0-3): "))
            excel = int(input("Nivell d'excel (0-3): "))
            formacio = int(input("Nivell de formació (0-3): "))

            if validar_dades(experiencia, catala, excel, formacio):
                return {
                    'nom': nom,
                    'telefon': telefon,
                    'experiencia': experiencia,
                    'catala': catala,
                    'excel': excel,
                    'formació': formacio
                }
            else:
                print("Error: Valors fora de rang permès. Torna a introduir les dades d'aquest candidat.")
                return None
        except ValueError:
            print("Error: Has d'introduir un número enter vàlid.")
            return None
```

Aquesta funció és calcular_puntuacio(), que calcula la puntuació total del candidat multiplicant els anys d'experiència per 1.5 i sumant-hi els nivells obtinguts en català, excel i formació.

```python
def calcular_puntuacio(candidat):
  puntuacio = (candidat['experiencia'] * 1.5) + candidat['catala'] + candidat['excel'] + candidat['formació']
  return puntuacio
```

Aquesta funció és classificar_candidat(), que avalua la puntuació obtinguda mitjançant estructures condicionals (if-elif-else) per assignar una categoria: "Prioritat Alta" si és major o igual a 12, "Candidat adequat" si és major o igual a 8, o "No prioritari" en cas contrari.

```python
def classificar_candidat(puntuacio):
    if puntuacio >= 12:
        return "Prioritat Alta"
    elif puntuacio >= 8:
        return "Candidat adequat"
    else:
        return "No prioritari"
```

Aquesta funció és mostrar_individual(), que s'encarrega d'imprimir per pantalla de manera formatada el nom del candidat, la puntuació obtinguda i la classificació final assignada.

```python
def mostrar_individual(candidat, puntuacio, classificacio):
    print(f"Candidat/a: {candidat['nom']}")
    print(f"Puntuació obtinguda: {puntuacio} punts")
    print(f"Classificació: {classificacio}")
```

Aquesta funció és mostrar_informe(), que genera un informe final de Recursos Humans. Primer defineix un diccionari d'ordre de prioritat i utilitza la funció sorted() combinada amb una funció lambda per ordenar els candidats segons la seva categoria i de forma descendent per la puntuació. Després, recorre la llista ordenada amb un bucle per mostrar un llistat numerat amb tots els detalls.

```python
def mostrar_informe(candidats_complets):
    ordre_prioritat = {"Prioritat Alta": 0, "Candidat adequat": 1, "No prioritari": 2}

    candidats_ordenats = sorted(
        candidats_complets,
        key=lambda candidat: (ordre_prioritat[candidat['classificacio']], -candidat['puntuacio'])
    )

    print("\n" + "="*45)
    print("                 INFORME FINAL RRHH")
    print("="*45)

    numero = 1
    for candidat in candidats_ordenats:
        print(f"{numero}. [{candidat['classificacio']}] {candidat['nom']} - Tel: {candidat['telefon']} - Punts: {candidat['puntuacio']}")
        numero += 1
    print("="*45)
```

Per finalitzar, la funció main_seleccio() integra tot el procés de selecció establint un màxim de 10 candidats vàlids. Utilitza un bucle while per gestionar les inscripcions, comprova si hi ha errors en les dades retornades, calcula la puntuació i classificació de cada candidat vàlid, afegeix la informació a una llista i finalment crida a l'informe final.

```python
def main_seleccio():
    max_candidats = 10
    candidats_complets = []
    inscrits_valids = 0

    print("Iniciant procés de selecció (màxim 10 candidats vàlids)...")

    while inscrits_valids < max_candidats:
        dades = obtenir_dades(inscrits_valids + 1)

        if dades is None:
            print("Avís: Inscripció descartada per errors en les dades. Es passa al següent intent.\n")
            continue

        puntuacio = calcular_puntuacio(dades)
        classificacio = classificar_candidat(puntuacio)

        dades['puntuacio'] = puntuacio
        dades['classificacio'] = classificacio

        candidats_complets.append(dades)
        mostrar_individual(dades, puntuacio, classificacio)

        inscrits_valids += 1

    mostrar_informe(candidats_complets)
```

```python
main_seleccio()
```
