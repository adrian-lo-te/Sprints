```python
import pandas as pd
```

La primera funció és normalitzar(). Utilitzo .strip() per eliminar els espais en blanc, .lower() per passar tot el text a minúscules i, per últim, .replace() per convertir els espais interns en guions baixos.

```python
def normalitzar(sintoma):
  return sintoma.strip().lower().replace(" ", "_")
```

La següent funció és transformar(), que rep la llista de registres en brut i la neteja. Primer he creat una llista buida i amb bucle for recorr cada registre. Dins d'aquest bucle, utilitz .split('|') per separar la cadena en tres parts. La part dels símptomes es torna a dividir utilitzant la coma com a separador, i s'aplica la funció normalitzar a cadascun dels elements. Per últim, he creat un diccionari dins del bucle, que es va afegint per cada pacient a la llista que he creat abans.

```python
def transformar(pacients):
  llista_pacients = []
  for pacient in pacients:
    parts = pacient.split('|')
    nom = parts[0].strip()
    edat = int(parts[1].strip())
    sintomes_bruts = parts[2].split(',')
    sintomes_nets = [normalitzar(s) for s in sintomes_bruts]
    diccionari_pacient = {
          "nom": nom,
          "edat": edat,
          "sintomes": sintomes_nets
      }
    llista_pacients.append(diccionari_pacient)
  return llista_pacients
```

Aquesta funció rep la llista de pacients nets juntament amb el diccionari de nivells de risc. Amb bucle for recorro els pacients de la llista, amb .get()agafo els valors del diccionari de nivells de risc, i amb un altre bucle for recorro els símptomes de cada pacient. Després, es sumen els nivells per crear els valors d'un nou paràmetre al diccionari. I per finalitzar es fica tot això en una nova llista, que és el que es retorna.

```python
def calcular_risc(llista_pacients, diccionari_risc):
  pacients_amb_risc = []
  for pacient in llista_pacients:
    risc_total = sum(
        diccionari_risc.get(sintoma, 0) for sintoma in pacient["sintomes"]
    )
    pacient_copia = pacient.copy()
    pacient_copia["risc_total"] = risc_total
    pacients_amb_risc.append(pacient_copia)
  return pacients_amb_risc
```

Aquesta funció transforma en un data frame la llista que retorna l'anterior funció i ordena per la columna risc_total.

```python
def prioritzar(pacients_amb_risc):
  taula_pacients = pd.DataFrame(pacients_amb_risc)
  taula_final = taula_pacients.sort_values(by = "risc_total", ascending = False)
  return taula_final
```

Per finalitzar, aquesta funció integra totes les anteriors.

```python
def processar(pacients, risc):
  llista_trans = transformar(pacients)
  taula_risc = calcular_risc(llista_trans, risc)
  taula = prioritzar(taula_risc)
  display(taula)
```

```python
pacients_brut = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
processar(pacients_brut, nivells_risc)
```
